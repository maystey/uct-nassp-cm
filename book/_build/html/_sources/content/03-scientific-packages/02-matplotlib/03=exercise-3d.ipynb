## Question 1

Consider a conical spiral defined by:

\begin{align*}
x &= s \cos(s \pi)\\
y &= s \sin(s \pi)\\
z &= s
\end{align*}

````python
s = np.linspace(0, 20, 1000)

x = s * np.cos(np.pi * s)
y = s * np.sin(np.pi * s)
z = s


fig, ax = plt.subplots(subplot_kw=dict(projection='3d'), figsize = (10, 8))

ax.plot(x, y, z, color = 'k')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

plt.show()
fig.clear()
````

## Question 2

The simplest example of a 2D Gaussian can be defined by:

$$
f(x, y) = A \exp\left( -\left(\frac{\left(x - x_0\right)^2}{2 \sigma_x^2} + \frac{\left(y - y_0\right)^2}{2 \sigma_y^2} \right) \right) 
$$

where $A$ is the height of the Gaussian, $\exp()$ is the exponential function $e^{()}$, $(x_0, y_0)$ is the position of the peak of the Gaussian, and $\sigma_x$ and $\sigma_y$ determine the spread of the Gaussian along the $x$ and $y$ axis respectively.

Plot a surface defined by:

$$
z = f(x, y)
$$

For a Gaussian with

- $A = 1$
- $x_0 = y_0 = 0$
- $\sigma_x = 1$
- $\sigma_y = 2$

as illustrated below (you need not match the illustration):

````python
A = 1

x0 = 0
sigma_x = 1

y0 = 0
sigma_y = 2

x = np.linspace(-4, 4, 40)
y = np.linspace(-6, 6, 40)

X, Y = np.meshgrid(x, y)

X_diff = X - x0
Y_diff = Y - y0

Z = A * np.exp( -0.5 * (X_diff * X_diff / sigma_x / sigma_x + Y_diff * Y_diff / sigma_y / sigma_y) )

fig, ax = plt.subplots(subplot_kw=dict(projection='3d'), figsize = (8, 6))

ax.plot_surface(X, Y, Z, color = '0.8', edgecolor = 'k')

ax.view_init(50)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')

plt.show()
fig.clear()
````
