## Question 1

NumPy has a number of [Mathematical functions](https://numpy.org/doc/stable/reference/routines.math.html) and [Constants](https://numpy.org/doc/stable/reference/constants.html). Use these to plot the following:

(1) $sin(x)$  for  $0 < x < 2\pi$

```python
x1 = np.linspace(0, 2 * np.pi, 1000)

fig1, ax1 = plt.subplots()

ax1.plot(x1, np.sin(x1), 'k-')
ax1.set_xlabel('x')
ax1.set_ylabel('y')

plt.show()

fig1.clear()
```

(2) $|x - 2|$  for  $0 < x < 3$

```python
x2 = np.linspace(0, 3, 1000)

fig2, ax2 = plt.subplots()

ax2.plot(x2, np.abs(x2 - 2), 'k-')
ax2.set_xlabel('x')
ax2.set_ylabel('y')

plt.show()
fig2.clear()
```

Make sure to label the $x$ and $y$ axis.

## Question 2

Plot a unit circle. Remember that the circle can be parameterized as:

\begin{align}
x(s) &= \cos(2 \pi s)\\
y(s) &= \sin(2 \pi s)
\end{align}

for $0 < s < 1$.

```python
s = np.linspace(0, 1, 1000)

x,y = np.cos(2 * np.pi * s), np.sin(2 * np.pi * s)

fig_circ, ax_circ = plt.subplots()

ax_circ.plot(x, y, 'k-')

ax_circ.set_aspect('equal')
ax_circ.set_xlabel('x')
ax_circ.set_ylabel('y')

plt.show()
fig_circ.clear()
```

**Hint**: You can use the [`set_aspect()`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.set_aspect.html) method of the axis object to prevent the circle from distorting.

## Question 3

Plot a spiral, parameterized by:

\begin{align}
x(s) &= s \cos(2 \pi s)\\
y(s) &= s \sin(2 \pi s)
\end{align}

for $0 < s < 1$.

```python
s = np.linspace(0, 1, 1000)

x,y = np.cos(2 * np.pi * s), np.sin(2 * np.pi * s)

fig_spiral, ax_spiral = plt.subplots()

ax_spiral.plot(s * x, s * y, 'k-')

ax_spiral.set_aspect('equal')
ax_spiral.set_xlabel('x')
ax_spiral.set_ylabel('y')

plt.show()
fig_spiral.clear()
```
