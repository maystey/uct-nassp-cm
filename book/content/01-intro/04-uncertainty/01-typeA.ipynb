\begin{frame}{Type A Uncertainty Evaluation}{Best Approximation of the Expectation Value}
    \begin{itemize}
        \item Measurand $X$, each time we make an observation/measurement ($x_i$), there is \textbf{random variation}.
        \item In most cases, the \textbf{best approximation of the expectation value} of $x$ ($\mu_x$) is the \textbf{arithmetic mean} $\bar{x}$
        \item Given $n$ independent observations under the same conditions (the more the better):

        \begin{equation*}
            \bar{x} = \frac{1}{n} \sum_{k-1}^{n} x_k
        \end{equation*}
    \end{itemize}
\end{frame}

\begin{frame}{Type A Uncertainty Evaluation}{Best Approximation of the Uncertainty}
    \begin{itemize}
        \item The best approximation of the variance of the PDF for $x$ ($\sigma^2$) is given by the \textbf{experimental variance}:
        
        \begin{equation*}
            s^2(x_k) = \frac{1}{n-1} \sum_{j=1}^{n} (x_k - \bar{x})^2 
        \end{equation*}

        using \texttt{numpy.std()}, be sure to set the keyword argument \texttt{ddof=1}.

        \item The variance of $\bar{x}$ is given by $\sigma^2(\bar{x}) = \sigma^2 / n$, the best estimate for this is given by:
        
        \begin{equation*}
            s^2(\bar{x}) = \frac{s^2(x_k)}{n}
        \end{equation*}

        \item The \textbf{experimental standard deviation of the mean} ($s(\bar{x})$) quantifies how well $\bar{x}$ estimates the expectation value $\mu_x$ of $x$.
        \item The standard uncertainty of $x$ is thus quoted as $u(x) = s(\bar{x})$.
    \end{itemize}
\end{frame}
