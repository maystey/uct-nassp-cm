\begin{frame}{Type B Uncertainty Evaluation}
    \begin{itemize}
        \item Estimate of a quantity not obtained from repeated observations
        \item The standard uncertainty is determined by ``scientific judgement'' based on information available:
        \begin{itemize}
            \item previous measurement data
            \item  experience with or general knowledge of the behaviour and properties of relevant materials and instruments
            \item manufacturer's specifications
            \item data provided in calibration and other certificates;
            \item uncertainties assigned to reference data taken from handbooks.
        \end{itemize}
    \item Even more than for Type A, there is no ``one-size-fits-all'' solution.
    \end{itemize}
\end{frame}

\begin{frame}{Type B Uncertainty Evaluation}{Quoting Uncertainties from Other Sources}

\begin{itemize}
    \item The source you are quoting from didn't need to follow the GUM framework. If it is clear enough how the uncertainty was obtained, it can be converted to the GUM framework.
    \item This conversion must be performed on a case-by-case basis\autocite{JCGMGUM}.
\end{itemize}

\end{frame}

\begin{frame}{Type B Uncertainty Evaluation}{Truncation Error}

\begin{itemize}
    \item Uncertainty from truncation errors mostly have to be evaluated as Type B uncertainties.
    \item For many of the methods in this module it is not possible to determine the distribution of likely values within error bounds.
    \item See the next 2 slides on how to deal with this lack of information.
\end{itemize}

\end{frame}
