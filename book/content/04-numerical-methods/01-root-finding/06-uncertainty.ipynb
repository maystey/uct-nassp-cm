# {docnum}`step ref` Uncertainty in Numerical Root Finding

\begin{frame}{Root Finding}
\begin{itemize}
    \item Root finding methods to solve $f(x) = 0$.
    \item Find $x$ - best approximation of the root:
    \begin{itemize}
        \item Bisection method: midpoint of bracket for last iteration
        \item Secant and Newton-Raphson: root estimate of last iteration
    \end{itemize}
    \item Root finding methods discussed use tolerance as stopping condition.
    \item This tolerance defines the limits of an interval in which we know the root is contained.
    \item Interval given by $[x - \epsilon, x + \epsilon]$ for:
    \begin{itemize}
        \item Bisection method with tolerance of $2\epsilon$
        \item Secant or Newton-Raphson method with tolerance of $\epsilon$ 
    \end{itemize}
    \item  We don't know more about where the root actually is in this interval: Type B uncertainty analysis using rectangular PDF.
    \item Choose tolerance such that uncertainty from root finding is negligible when compared to other sources.
\end{itemize}
\end{frame}
