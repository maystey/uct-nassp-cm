# {docnum}`step ref` Improved Trapezoidal Rule

We can use the theoretical error of the composite trapezoidal rule to improve our approximation of the integral.

We presented an expression of the composite trapezoidal rule in Equation {eq}`trapz:compound_error`.
This was the result of applying the Discrete Average Value theorem {cite}`nm-i-t-epperson-numerical-methods` to

$$
I_n - T_n(f) = - \frac{1}{12} h^3 \sum_{i=0}^{n-1} f''(\xi_i)
$$(imp_trapz:error)

for $\xi_i \in [a+ih, a+(i+1)h]$.
Equation {eq}`imp_trapz:error` is a direct result of summing the errors of the trapezoidal rule for each sub-interval.

Note that the sum of $f''$ in {eq}`imp_trapz:error` can be interpreted as a (finite) Riemann sum:

$$
\sum_{i=0}^{n-1} h f''(\xi_i) \approx I(f'') = f'(b) - f'(a)
$$

which can be substituted into {eq}`imp_trapz:error`:

$$
I_n - T_n(f) \approx - \frac{1}{12} h^2 (f'(b) - f'(a))
$$(imp_trapz:error_approx)

Equation {eq}`imp_trapz:error_approx` can be used to define the improved trapezoidal method:

$$
T_n^C(f) = T_n(f) - \frac{1}{12} h^2 (f'(b) - f'(a))
$$(imp_trapz:improved)

Although we don't have a way of estimating the error of {eq}`imp_trapz:improved` directly, we can use {eq}`imp_trapz:error_approx` as an over-estimation.

Equation {eq}`imp_trapz:error_approx` can also be used to approximate the number of sub-intervals required to approximate the integral using the (improved) trapezoidal rule within a given tolerance.
If we have a desired tolerance for a method, we require that the error is smaller than this:

\begin{align*}
|I_n(f) - T_n(f)| &< \text{tolerance}\\
\therefore \frac{1}{12}\frac{(b-a)^2}{n^2} |f'(b) - f'(a)| &< \text{tolerance}\\
\therefore n &> \frac{\sqrt{3}}{6} |b - a| \sqrt{\frac{|f'(b) - f'(a)|}{\text{tolerance}}}
\end{align*}
