# {docnum}`step ref` Comparing Quantities with Uncertainties

Using expanded uncertainties, two best-approximations of a measurand with associated standard uncertainties can be compared.
Two approximations are said to be in agreement with the given confidence level if their expanded uncertainty intervals overlap (note that the confidence level for both expanded uncertainties must be identical).
