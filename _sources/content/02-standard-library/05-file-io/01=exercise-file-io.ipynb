# Exercise {docnum}`ref`

Write a script that can read a text file and count:

- The number of characters in the text
- The number of words
- The number of vowels

Test your script on {download}`alice_in_wonderland-lewis_carroll-chapter_5.txt<./exercise-data/alice_in_wonderland-lewis_carroll-chapter_5.txt>`, an extract from Lewis Carroll's Alice in Wonderland, where you should get something along the lines of:
 - 11990 characters
 - 2439 words
 - 3440 vowels
depending on how you define certain criteria.
