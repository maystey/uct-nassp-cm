# {docnum}`step ref` Tuple

Just as strings are a sequence of characters, tuples are a sequence of objects. This makes their use far more general.

You can define tuples by using brackets and separating objects by commas:

```python
t = ('a', 1, 'b', 2, 'c', 3)
```

Like strings, tuples can be indexed and sliced:

```python
print('Index 3:', t[3])
print('Slice from index 3:', t[3:])
```

Tuples are also immutable (like strings):

```python
t[2] = 5
```

If you need to define a tuple with a single element, then you simply place a trailing comma:

```python
t = (1,)

print(type(t))
print(len(t))
```

## Tuple Packing

You can create a tuple without using brackets:

```python
t = 1, 2, 3, 'a', 'b', 'c'

print(t)
```

This is called tuple packing, and is often more convenient than including the brackets. Note that brackets are necessary if you want to define a tuple while passing it as an argument to a function, and other one-line use-cases.

## Tuple Unpacking

You can unpack a tuple into multiple variables, just like you can pack multiple values into a tuple:

```python
t = (1, 2, 3)
print('t is ', t)

x, y, z = t
print('x is', x) 
print('y is', y)
print('z is', z)
```

You can nest tuple unpacking if you have a tuple in a tuple, for example:

```python
a, (b, c) = (1, (2, 3))

print('a is ', a)
print('b is ', b)
print('c is ', c)
```
