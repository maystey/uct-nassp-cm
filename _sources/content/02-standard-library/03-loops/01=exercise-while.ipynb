## Question 1

Consider the following series:

\begin{align*}
T_0 = 1\\
T_n = \left(T_{n - 1}\right)~^{3/2} + 2
\end{align*}

calculate the value of $n$ where $T_n$ first exceeds a value of 100.

**Hint:** You may wish to use a `while` loop.

(The answer is $n = 4$)
